### Research question

Which observable page-level signals can help prioritize content pages for review, and how can those signals be combined into a ranking that supports practical refresh decisions?

The decision is which pages should be reviewed first. A content or SEO practitioner could use the ranking to decide where to spend limited review and optimization time. A wrong recommendation has two main costs: reviewing a page that does not need attention wastes time, while missing a page that may need attention can delay useful improvements. Therefore, the goal is to rank pages for review and support decision-making, not to claim that the model proves why a page changed.

### What the data shows

The starter dataset contains 30,000 anonymized content records across 44 columns. The largest trend group is `down`, with 16,262 records, followed by `stable` with 5,962 and `up` with 4,388. The correlation between search volume and 90-day impressions is 0.001, which is very close to zero in this dataset. The median word count is 2,909 for pages in the `down` group and 2,848 for pages in the `up` group.

These numbers suggest that search volume alone does not have a strong linear relationship with impressions in this dataset, and that content characteristics such as word count can vary across trend groups. This makes it useful to investigate multiple observable signals together when prioritizing pages for review.

These are observations from the provided anonymized dataset and do not establish causal relationships or explain Google's ranking algorithm.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.



import pandas as pd
import os

print("Current folder:", os.getcwd())
print("Files here:", os.listdir())



if not os.path.exists("flyrank-ml-internship"):
    !git clone https://github.com/vidhipandey26/flyrank-ml-internship.git
df = pd.read_csv(
    "flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"
)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nTrend direction:")
print(df["trend_direction"].value_counts())

print("\nCorrelation: search volume vs impressions:")
print(round(df["search_volume"].corr(df["impressions_90d"]), 3))

print("\nMedian word count by trend direction:")
print(df.groupby("trend_direction")["word_count"].median().round(0))
```
